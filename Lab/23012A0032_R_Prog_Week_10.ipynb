## 1. String substitution

```r
y <- "Number of participants: 25"
sub("25", "30", y)
y <- "Mr. Singh is the smart one. Mr. Singh is funny, too."
sub("Mr. Singh", "Professor Jha", y)
gsub("Mr. Singh", "Professor Jha", y)
```

## 2. Searching strings with grep

```r
str <- c("R Course", "exercises", "include examples of R language")
grep("ex", str, value = TRUE)
grep("ex", str, value = FALSE)
str2 <- c("R Course", "exercises", "include examples of r language", "in R software.")
grep("R", str2, ignore.case = FALSE, value = TRUE)
grep("R", str2, ignore.case = TRUE, value = TRUE)
x <- "R course 24.07.2021"
y <- "Number of participants: 25"
grep("our", c(x, y))
grep("Num", c(x, y))
```

## 3. Searching strings with grepl

```r
str <- c("R Course", "exercises", "include examples of R language")
grepl("R", str)
grepl("ex", str)
```

## 4. Creating and inspecting data frames

```r
students <- data.frame(
  Name = c("Asha", "Riya", "Neha", "Amit"),
  Age = c(21, 20, 22, 21),
  Marks = c(85, 92, 76, 88),
  Department = c("CMPN", "IT", "CMPN", "EXTC")
)
students
str(students)
colnames(students)
rownames(students)
summary(students)
is.numeric(students$Age)
is.factor(students$Department)
```

## 5. Row names and column access

```r
painters <- data.frame(
  Composition = c(10, 15, 8, 12, 0),
  Drawing = c(8, 16, 13, 16, 15),
  Colour = c(16, 4, 16, 9, 8),
  Expression = c(3, 14, 7, 8, 0),
  School = factor(c("A", "A", "A", "A", "A")),
  row.names = c("Da Udine", "Da Vinci", "Del Piombo", "Del Sarto", "Fr. Penni")
)
painters
painters$Drawing
painters["School"]
rownames(painters)
colnames(painters)
summary(painters)
summary(painters$School)
```

## 6. Subsetting data frames

```r
subset(painters, School == "A")
subset(painters, Composition <= 10)
subset(painters, School == "A", select = c(-3, -5))
painters[painters$Composition <= 10, ]
painters[, c("Composition", "Drawing", "Expression")]
```

## 7. Attaching a data frame

```r
attach(painters)
Drawing
School
detach(painters)
```

## 8. Splitting data frames

```r
split(painters, painters$School)
```

## 9. Combining data frames with cbind

```r
df1 <- data.frame(
  state = c("UP", "MP", "AP", "JK"),
  popnsize = c(1000, 2000, 3000, 4000)
)
df2 <- data.frame(
  state = c("UP", "MP", "AP", "JK"),
  samplesize = c(100, 200, 300, 400),
  surveycompleted = c("Yes", "No", "Yes", "No")
)
cbind(df1, df2)
```

## 10. Merging data frames

```r
merge(df1, df2, by = "state")
```

## 11. Combining data frames with rbind

```r
df11 <- data.frame(
  state = c("UP", "MP", "AP", "JK"),
  popnsize = c(1000, 2000, 3000, 4000)
)
df22 <- data.frame(
  state = c("Bihar", "Delhi", "Punjab"),
  popnsize = c(100, 200, 300)
)
rbind(df11, df22)
```
